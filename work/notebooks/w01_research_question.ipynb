## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python

import pandas as pd

# Load the official starter dataset
url = "https://raw.githubusercontent.com/mohamedmorsy-ai/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

# Check the required columns
required = [
    "content_id",
    "impressions_90d",
    "content_age_days",
    "trend_direction"
]

assert all(col in df.columns for col in required), "Required columns missing"

# Prepare the data
df["impressions_90d"] = pd.to_numeric(
    df["impressions_90d"], errors="coerce"
)

df["content_age_days"] = pd.to_numeric(
    df["content_age_days"], errors="coerce"
)

# Keep eligible content items
eligible = df[
    (df["impressions_90d"] > 0) &
    (df["content_age_days"] >= 90)
].drop_duplicates(subset="content_id")

# Identify declining content
declining = eligible[
    eligible["trend_direction"].astype(str).str.lower().str.strip() == "down"
]

# Calculate three real statistics
total_pages = len(eligible)
declining_pages = len(declining)
declining_impressions = declining["impressions_90d"].sum()

decline_pct = (
    declining_pages / total_pages * 100
    if total_pages > 0 else 0
)

print("ML-02: Starter Data Exploration")
print("-" * 45)

print(f"1. Eligible content pages: {total_pages:,}")
print(f"2. Declining pages: {declining_pages:,} ({decline_pct:.2f}%)")
print(f"3. 90-day impressions on declining pages: {declining_impressions:,.0f}")

print("\nInterpretation:")
print(
    "These statistics describe the scale of the content "
    "review problem and support further investigation."
)
print(
    "They do not prove that refreshing a page will "
    "improve its search performance."
)
```
